```python
import numpy as np
import meshio
import meshplot as mp
```

```python
def load_off(fname):
    mesh = meshio.read(fname)
    V = np.asarray(mesh.points)
    F = np.asarray(mesh.cells_dict["triangle"])
    return V, F
```

```python
def plot_two_true(file1, file2, shift=0.8):
    V1, F1 = load_off(file1)
    V2, F2 = load_off(file2)

    V1 = V1.copy()
    V2 = V2.copy()

    # centrar horizontalmente
    V1[:, 0] -= V1[:, 0].mean()
    V2[:, 0] -= V2[:, 0].mean()

    # alinear verticalmente por la parte más baja
    V1[:, 1] -= V1[:, 1].min()
    V2[:, 1] -= V2[:, 1].min()

    # separar un poco en horizontal
    V1[:, 0] -= shift / 2
    V2[:, 0] += shift / 2

    color1 = np.tile(np.array([[0.95, 0.95, 0.95]]), (V1.shape[0], 1))
    color2 = np.tile(np.array([[0.95, 0.95, 0.95]]), (V2.shape[0], 1))

    p = mp.plot(V1, F1, c=color1)
    p.add_mesh(V2, F2, c=color2)

    return p
```

```python
def center_mesh(V):
    return V - V.mean(axis=0)

def add_mesh_at(p, V, F, shift_x=0.0, shift_y=0.0, color_val=0.95):
    V_new = center_mesh(V).copy()
    V_new[:, 0] += shift_x
    V_new[:, 1] += shift_y
    color = np.tile(np.array([[color_val, color_val, color_val]]), (V_new.shape[0], 1))
    p.add_mesh(V_new, F, c=color)

files = [
    ("horse_40_v0_test_true.off",   -0.8,  1.2),
    ("horse_44_v0_test_true.off",    0.8,  1.2),
    ("camel_40_v0_test_true.off",   -0.8,  0.0),
    ("camel_44_v0_test_true.off",    0.8,  0.0),
    ("elephant_40_v0_test_true.off",-0.8, -1.2),
    ("elephant_44_v0_test_true.off", 0.8, -1.2),
]

#first mesh.
V0, F0 = load_off(files[0][0])
V0 = center_mesh(V0)
V0[:, 0] += files[0][1]
V0[:, 1] += files[0][2]
color0 = np.tile(np.array([[0.95, 0.95, 0.95]]), (V0.shape[0], 1))

p = mp.plot(V0, F0, c=color0)

#rest of meshes.
for fname, sx, sy in files[1:]:
    V, F = load_off(fname)
    add_mesh_at(p, V, F, shift_x=sx, shift_y=sy, color_val=0.95)

p
```

```python
p.save("GALLOP.png")
```

```python
p = plot_two_true(
    "camel_40_v0_test_true.off",
    "camel_44_v0_test_true.off"
)

p
```
